```python
import pandas as pd
```

```python
data=pd.read_excel("./coffeeOrdersData.xlsx",sheet_name=[0,1,2])
```

```python
orders=data[0]
costumers=data[1]
products=data[2]
```

```python
orders.head(1)
```

```python
ordersdf=orders.copy()
```

```python
ordersdf.head(1)
```

```python
ordersdf.drop(['Customer Name', 'Email', 'Country', 'Coffee Type', 'Roast Type',
       'Size', 'Unit Price', 'Sales'],axis=1,inplace=True)
```

```python
ordersdf.columns
```

```python
ordersdf=ordersdf.merge(costumers,how='left',on='Customer ID')
```

```python
ordersdf.drop(['Phone Number','Address Line 1','City','Postcode','Loyalty Card'],axis=1,inplace=True)
```

```python
ordersdf
```

```python
orders.head(2)
```

```python
ordersdf=ordersdf.merge(products,how='left',on='Product ID')
```

```python
ordersdf.drop(['Price per 100g'],axis=1,inplace=True)
```

```python
ordersdf.head()
```

```python
ordersdf.rename(columns={"Profit":"Sales"},inplace=True)
```

```python
ordersdf.head(1)
```

```python
ordersdf.fillna("",inplace=True)
```

```python
ordersdf['Sales']=(ordersdf['Sales']*ordersdf['Quantity'])
```

```python
ordersdf['Coffe type Name']=ordersdf['Coffee Type'].map({'Rob':'Robusta','Exc':'Excelsa','Ara':'Arabica','Lib':'Librica'})
```

```python
ordersdf['Order Date']
```

```python
ordersdf['Roast Type']=ordersdf['Roast Type'].map({'L':'Lite','M':'Medium','D':'Dark'})
```

```python
ordersdf['Order Date'] = ordersdf['Order Date'].dt.strftime('%Y-%m')
```

```python
ordersdf['Order Date']
```

```python
ordersdf.duplicated().sum()
```

```python
ordersdf.head()
```

```python
prited=pd.pivot_table(ordersdf,values='Sales',columns=['Coffe type Name'],index='Order Date')
```

```python
import matplotlib.pyplot as plt
import seaborn as sbn
```

```python
# sbn.lineplot(prited)
prited.plot(kind='line')
plt.show()
```

```python
ordersdf.groupby("Coffee Type").size().plot(kind="bar")
plt.show()
```

```python
sbn.barplot(ordersdf['Country'].value_counts())
```

```python
ordersdf.groupby("Country")['Sales'].sum().sort_values(ascending=False).plot(kind='bar')
plt.show()
```

```python
prited=pd.pivot_table(ordersdf,values='Sales',columns=['Coffe type Name'],index='Country')
```

```python
# sbn.barplot(prited,legend=True)
prited.plot(kind='bar')
plt.title("sales by country")
plt.show()
```

```python
ordersdf.groupby('Order Date')['Sales'].sum().plot(kind='line')
plt.xlabel("time")
plt.ylabel("sales")
plt.title("coffe sales by time")
plt.show()
```
